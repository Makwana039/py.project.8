```python
# --------------------------
# Analyzer
# --------------------------
import numpy as np


class DataAnalytics:
    def __init__(self):
        self.__array = None

    # 1. Create arrays
    def create_array(self):
        print("\nChoose Array Type:")
        print("1. 1D Array")
        print("2. 2D Array")
        print("3. 3D Array")

        try:
            choice = int(input("Enter your choice: "))

            if choice == 1:
                n = int(input("Enter number of elements: "))
                values = list(map(
                    float,
                    input(f"Enter {n} elements separated by space: ").split()
                ))
                if n <= 0 or len(values) != n:
                    print("Invalid number of elements.")
                    return
                self.__array = np.array(values)

            elif choice == 2:
                rows = int(input("Enter number of rows: "))
                cols = int(input("Enter number of columns: "))

                if rows <= 0 or cols <= 0:
                    print("Dimensions must be positive.")
                    return

                values = list(map(
                    float,
                    input(f"Enter {rows * cols} elements separated by space: ").split()
                ))

                if len(values) != rows * cols:
                    print("Invalid number of elements.")
                    return

                self.__array = np.array(values).reshape(rows, cols)

            elif choice == 3:
                depth = int(input("Enter depth: "))
                rows = int(input("Enter number of rows: "))
                cols = int(input("Enter number of columns: "))

                if min(depth, rows, cols) <= 0:
                    print("Dimensions must be positive.")
                    return

                total = depth * rows * cols
                values = list(map(
                    float,
                    input(f"Enter {total} elements separated by space: ").split()
                ))

                if len(values) != total:
                    print("Invalid number of elements.")
                    return

                self.__array = np.array(values).reshape(
                    depth, rows, cols
                )

            else:
                print("Invalid choice.")
                return

            print("\nArray created successfully!")
            print(self.__array)
            print("Shape:", self.__array.shape)
            print("Dimensions:", self.__array.ndim)

        except ValueError:
            print("Please enter valid numeric values.")

    # 2. Mathematical operations
    def mathematical_operations(self):
        if not self._check_array():
            return

        print("\n1. Addition")
        print("2. Subtraction")
        print("3. Multiplication")
        print("4. Division")
        print("5. Dot Product")
        print("6. Matrix Multiplication")

        try:
            choice = int(input("Enter your choice: "))

            if choice in (1, 2, 3, 4):
                value = float(input("Enter a number: "))

                if choice == 1:
                    result = self.__array + value
                elif choice == 2:
                    result = self.__array - value
                elif choice == 3:
                    result = self.__array * value
                else:
                    if value == 0:
                        print("Cannot divide by zero.")
                        return
                    result = self.__array / value

            elif choice in (5, 6):
                if self.__array.ndim != 2:
                    print("This operation requires a 2D array.")
                    return

                rows = int(input("Enter rows of second array: "))
                cols = int(input("Enter columns of second array: "))

                if rows <= 0 or cols <= 0:
                    print("Dimensions must be positive.")
                    return

                values = list(map(
                    float,
                    input(f"Enter {rows * cols} elements: ").split()
                ))

                if len(values) != rows * cols:
                    print("Invalid number of elements.")
                    return

                second = np.array(values).reshape(rows, cols)

                if choice == 5:
                    if self.__array.size != second.size:
                        print("Arrays must have equal numbers of elements.")
                        return
                    result = np.dot(
                        self.__array.ravel(), second.ravel()
                    )
                else:
                    if self.__array.shape[1] != second.shape[0]:
                        print("Incompatible matrix dimensions.")
                        return
                    result = np.matmul(self.__array, second)

            else:
                print("Invalid choice.")
                return

            print("Result:\n", result)

        except ValueError:
            print("Invalid input.")

    # 3. Combine or split arrays
    def combine_split(self):
        if not self._check_array():
            return

        print("\n1. Combine Arrays")
        print("2. Split Array")

        try:
            choice = int(input("Enter your choice: "))

            if choice == 1:
                if self.__array.ndim not in (1, 2):
                    print("Combine supports 1D and 2D arrays here.")
                    return

                values = list(map(
                    float,
                    input("Enter elements for the second array: ").split()
                ))
                second = np.array(values)

                if self.__array.ndim == 1:
                    result = np.concatenate((self.__array, second))
                else:
                    if second.size != self.__array.shape[1]:
                        print(
                            "Enter the same number of elements as "
                            "the columns in the current array."
                        )
                        return
                    second = second.reshape(1, -1)
                    result = np.concatenate(
                        (self.__array, second), axis=0
                    )

                print("Combined Array:\n", result)

            elif choice == 2:
                axis = 0

                if self.__array.ndim > 1:
                    axis = int(input("Enter split axis (0 or 1): "))

                if axis < 0 or axis >= self.__array.ndim:
                    print("Invalid axis.")
                    return

                parts = int(input("Enter number of parts: "))

                if parts <= 0:
                    print("Number of parts must be positive.")
                    return

                result = np.array_split(
                    self.__array, parts, axis=axis
                )

                for i, part in enumerate(result, start=1):
                    print(f"Part {i}:\n{part}")

            else:
                print("Invalid choice.")

        except ValueError:
            print("Invalid input.")

    # 4. Search, sort, and filter
    def search_sort_filter(self):
        if not self._check_array():
            return

        print("\n1. Search")
        print("2. Sort")
        print("3. Filter")

        try:
            choice = int(input("Enter your choice: "))

            if choice == 1:
                value = float(input("Enter value to search: "))
                positions = np.argwhere(self.__array == value)

                if positions.size:
                    print("Value found at indices:\n", positions)
                else:
                    print("Value not found.")

            elif choice == 2:
                order = input(
                    "Enter A for ascending or D for descending: "
                ).strip().upper()

                if order == "A":
                    result = np.sort(self.__array, axis=None)
                elif order == "D":
                    result = np.sort(
                        self.__array, axis=None
                    )[::-1]
                else:
                    print("Invalid order.")
                    return

                print("Sorted Array:\n", result)

            elif choice == 3:
                value = float(input("Filter elements greater than: "))
                result = self.__array[self.__array > value]
                print("Filtered Elements:\n", result)

            else:
                print("Invalid choice.")

        except ValueError:
            print("Invalid input.")

    # 5. Aggregation and statistics
    def aggregates_statistics(self):
        if not self._check_array():
            return

        print("\n1. Sum")
        print("2. Mean")
        print("3. Median")
        print("4. Standard Deviation")
        print("5. Variance")
        print("6. Minimum and Maximum")
        print("7. Percentiles")
        print("8. Correlation Coefficient")

        try:
            choice = int(input("Enter your choice: "))
            arr = self.__array

            if choice == 1:
                print("Sum:", np.sum(arr))
            elif choice == 2:
                print("Mean:", np.mean(arr))
            elif choice == 3:
                print("Median:", np.median(arr))
            elif choice == 4:
                print("Standard Deviation:", np.std(arr))
            elif choice == 5:
                print("Variance:", np.var(arr))
            elif choice == 6:
                print("Minimum:", np.min(arr))
                print("Maximum:", np.max(arr))
            elif choice == 7:
                p = float(input("Enter percentile (0-100): "))
                if not 0 <= p <= 100:
                    print("Percentile must be between 0 and 100.")
                    return
                print(f"{p}th Percentile:", np.percentile(arr, p))
            elif choice == 8:
                if arr.ndim != 2 or arr.shape[1] < 2:
                    print("Use a 2D array with at least two columns.")
                    return
                print("Correlation Matrix:\n", np.corrcoef(arr, rowvar=False))
            else:
                print("Invalid choice.")

        except ValueError:
            print("Invalid input.")

    # Private helper method
    def _check_array(self):
        if self.__array is None:
            print("Please create an array first.")
            return False
        return True

    # Class method
    @classmethod
    def project_name(cls):
        return "NumPy Analyzer"

    # Static method
    @staticmethod
    def welcome():
        print("=" * 40)
        print("       WELCOME TO NUMPY ANALYZER")
        print("=" * 40)

    # Main menu
    def run(self):
        self.welcome()

        while True:
            print("\nChoose an option:")
            print("1. Create a NumPy Array")
            print("2. Perform Mathematical Operations")
            print("3. Combine or Split Arrays")
            print("4. Search, Sort, or Filter Arrays")
            print("5. Compute Aggregates and Statistics")
            print("6. Exit")

            try:
                choice = int(input("Enter your choice: "))

                if choice == 1:
                    self.create_array()
                elif choice == 2:
                    self.mathematical_operations()
                elif choice == 3:
                    self.combine_split()
                elif choice == 4:
                    self.search_sort_filter()
                elif choice == 5:
                    self.aggregates_statistics()
                elif choice == 6:
                    print("Thank you for using NumPy Analyzer!")
                    break
                else:
                    print("Please choose a number from 1 to 6.")

            except ValueError:
                print("Please enter a valid menu number.")


if __name__ == "__main__":
    print(DataAnalytics.project_name())
    analyzer = DataAnalytics()
    analyzer.run()
```
